```python
import pandas as pd

indexing_times = pd.read_csv('/home/sbasir/Thesis/Thesis/NIR/Experiments/scripts/sbert/Indexing_Times.csv')
```

```python
# Calculate the average time for each percentage
average_times = {
    "10%": sum(indexing_times["10%"]) / len(indexing_times["10%"]),
    "20%": sum(indexing_times["20%"]) / len(indexing_times["20%"]),
    "30%": sum(indexing_times["30%"]) / len(indexing_times["30%"]),
    "40%": sum(indexing_times["40%"]) / len(indexing_times["40%"]),
    "50%": sum(indexing_times["50%"]) / len(indexing_times["50%"]),
    "60%": sum(indexing_times["60%"]) / len(indexing_times["60%"]),
    "70%": sum(indexing_times["70%"]) / len(indexing_times["70%"]),
    "80%": sum(indexing_times["80%"]) / len(indexing_times["80%"]),
    "90%": sum(indexing_times["90%"]) / len(indexing_times["90%"]),
    "100%": sum(indexing_times["100%"]) / len(indexing_times["100%"]),
}
average_times
```

```python
import matplotlib.pyplot as plt

# Create a line graph
percentages = list(average_times.keys())
times = list(average_times.values())

plt.figure(figsize=(10, 6))
plt.plot(percentages, times, marker='o', linestyle='-', color='b')
plt.title('Average Time to Index Documents by Percentage - 1 million documents - SBERT')
plt.xlabel('Percentage Indexed')
plt.ylabel('Average Time (seconds)')
plt.grid(True)

# Show the plot
plt.show()
```
