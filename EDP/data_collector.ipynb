```python
import csv

def extract_dataset_ids(csv_filename):
    dataset_ids = []

    with open(csv_filename, 'r') as csvfile:
        csv_reader = csv.reader(csvfile)
        next(csv_reader)  # Skip header

        for row in csv_reader:
            dataset_name = row[0]
            dataset_id = dataset_name.split('_')[0]
            dataset_ids.append(dataset_id)

    return dataset_ids

# Usage
csv_filename = '/home/sbasir/Thesis/Thesis/EDP/europeana_datasets - europeana_datasets.csv'  # Replace with the actual path to your CSV file
dataset_ids = extract_dataset_ids(csv_filename)
print(dataset_ids)
```
